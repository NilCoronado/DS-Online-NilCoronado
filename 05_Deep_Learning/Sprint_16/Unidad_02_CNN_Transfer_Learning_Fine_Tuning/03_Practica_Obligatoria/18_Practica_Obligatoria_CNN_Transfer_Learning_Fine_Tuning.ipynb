```python
# ── IMPORTS ────────────────────────────────────────────────────────────────
import os
os.environ['KERAS_BACKEND'] = 'torch'

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import cv2
import warnings
warnings.filterwarnings('ignore')

import keras
from keras.applications import MobileNetV2
from keras.applications.mobilenet_v2 import preprocess_input
from keras.layers import GlobalAveragePooling2D, Dense, Dropout, RandomFlip, RandomRotation
from keras.callbacks import EarlyStopping
from sklearn.metrics import classification_report, ConfusionMatrixDisplay

print(f"Keras {keras.__version__} | backend: {keras.backend.backend()}")

# ============================================================
# 1. CARGA DEL DATASET
# Usamos cats/dogs de la Práctica Obligatoria de Unidad_01
# (Intel Landscapes no disponible localmente — mismo dataset
#  que usa el workout de esta unidad)
# ============================================================

DATA_BASE = "../../Unidad_01_CNN_Redes_Convolucionales/03_Practica_Obligatoria/data"
TRAIN_DIRS = [f"github_train_{i}" for i in range(4)]
TEST_DIR   = "github_test"
IM_SIZE    = 64   # Buena relación velocidad/calidad en CPU

CLASS_NAMES = {0: "Gato", 1: "Perro"}

def load_images_from_dir(directory, im_size):
    X, y = [], []
    for filename in sorted(os.listdir(directory)):
        if not filename.lower().endswith('.jpg'):
            continue
        label = 0 if filename.startswith('cat') else 1
        img = cv2.imread(os.path.join(directory, filename))
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        img = cv2.resize(img, (im_size, im_size))
        X.append(img); y.append(label)
    return X, y

X_list, y_list = [], []
for td in TRAIN_DIRS:
    Xp, yp = load_images_from_dir(os.path.join(DATA_BASE, td), IM_SIZE)
    X_list.extend(Xp); y_list.extend(yp)
X_test_list, y_test_list = load_images_from_dir(os.path.join(DATA_BASE, TEST_DIR), IM_SIZE)

X_train_raw = np.array(X_list,      dtype='uint8')
y_train      = np.array(y_list)
X_test_raw  = np.array(X_test_list, dtype='uint8')
y_test       = np.array(y_test_list)

print(f"Train: {X_train_raw.shape} | gatos={(y_train==0).sum()}, perros={(y_train==1).sum()}")
print(f"Test : {X_test_raw.shape}  | gatos={(y_test==0).sum()},  perros={(y_test==1).sum()}")

# ============================================================
# 2. VISUALIZACIÓN Y MINIEDA
# ============================================================

fig, axes = plt.subplots(2, 5, figsize=(14, 6))
for i, ax in enumerate(axes.flat):
    ax.imshow(X_train_raw[i]); ax.set_title(CLASS_NAMES[y_train[i]]); ax.axis('off')
fig.suptitle(f"Muestras — {IM_SIZE}×{IM_SIZE} px (cats/dogs dataset)", fontsize=13)
plt.tight_layout(); plt.show()

fig, ax = plt.subplots(figsize=(5, 4))
ax.bar(['Gato', 'Perro'], [(y_train==0).sum(), (y_train==1).sum()], color=['#2196F3','#FF5722'])
ax.set_title("Distribución de clases — Train")
plt.tight_layout(); plt.show()

# ============================================================
# 3. PREPROCESADO MobileNetV2 → escala [-1, 1]
# ============================================================

X_train = preprocess_input(X_train_raw.astype('float32'))
X_test  = preprocess_input(X_test_raw.astype('float32'))
print(f"Rango X_train: [{X_train.min():.1f}, {X_train.max():.1f}]")

# ── Helper: evaluar y mostrar reporte ───────────────────────
def report_model(probs, y_te, title):
    y_pred = (probs >= 0.5).astype(int)
    print(f"\n{'='*50}\n  {title}\n{'='*50}")
    print(classification_report(y_te, y_pred, target_names=['Gato', 'Perro']))
    fig, ax = plt.subplots(figsize=(5, 4))
    ConfusionMatrixDisplay.from_predictions(y_te, y_pred,
        display_labels=['Gato', 'Perro'], ax=ax, colorbar=False)
    ax.set_title(f"Matriz de Confusión\n{title}")
    plt.tight_layout(); plt.show()
    return (y_pred == y_te).mean()

# ============================================================
# 4. MODELO BASE PREENTRENADO — MobileNetV2
# ============================================================

base_model = MobileNetV2(input_shape=(IM_SIZE, IM_SIZE, 3),
                          include_top=False, weights='imagenet')
print(f"\nMobileNetV2 cargada: {len(base_model.layers)} capas")

# ============================================================
# 5. TRANSFER LEARNING
# Estrategia: extraer features con base congelada (una sola pasada,
# muy rápido en CPU) → entrenar cabeza Dense sobre esas features.
# ============================================================

base_model.trainable = False  # Congela TODOS los pesos

# Extractor de features: base + GlobalAveragePooling
inp_fe = keras.Input(shape=(IM_SIZE, IM_SIZE, 3))
feat_out = GlobalAveragePooling2D()(base_model(inp_fe, training=False))
feature_extractor = keras.Model(inp_fe, feat_out, name="MobileNetV2_Features")

print("\nExtrayendo features (pasada única, rápida)...")
F_train = feature_extractor.predict(X_train, batch_size=128, verbose=1)  # (4000, 1280)
F_test  = feature_extractor.predict(X_test,  batch_size=128, verbose=1)  # (1000, 1280)
print(f"Features shape: {F_train.shape}")

# Cabeza densa (máx 2 capas ocultas + salida, según enunciado)
head_tl = keras.Sequential([
    keras.Input(shape=(F_train.shape[1],)),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(1, activation='sigmoid')
], name="TL_Head")

head_tl.compile(optimizer=keras.optimizers.Adam(1e-3),
                loss='binary_crossentropy', metrics=['accuracy'])
head_tl.summary()

early_tl = EarlyStopping(monitor='val_loss', patience=4,
                          restore_best_weights=True, verbose=1)

history_tl = head_tl.fit(
    F_train, y_train,
    batch_size=64,
    epochs=25,
    validation_split=0.2,
    callbacks=[early_tl],
    verbose=1
)

# Historial TL
hist_tl = pd.DataFrame(history_tl.history)
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
hist_tl[['loss','val_loss']].plot(ax=axes[0])
axes[0].set_title("Transfer Learning — Loss"); axes[0].set_xlabel("Epoch")
hist_tl[['accuracy','val_accuracy']].plot(ax=axes[1])
axes[1].set_title("Transfer Learning — Accuracy"); axes[1].set_xlabel("Epoch")
plt.tight_layout(); plt.show()

# Evaluación TL
tl_loss, tl_acc = head_tl.evaluate(F_test, y_test, verbose=0)
print(f"\nTransfer Learning — Test Loss: {tl_loss:.4f} | Test Accuracy: {tl_acc:.4f}")
probs_tl = head_tl.predict(F_test, verbose=0).flatten()
acc_tl = report_model(probs_tl, y_test, "Transfer Learning (base congelada)")

# ============================================================
# 6. FINE TUNING
# Descongelamos últimas ~54 capas de MobileNetV2.
# LR muy bajo para no destruir pesos preentrenados.
# Inicializamos cabeza con los pesos del TL.
# ============================================================

FINE_TUNE_AT = 100  # Dejamos congeladas las primeras 100 capas
base_model.trainable = True
for layer in base_model.layers[:FINE_TUNE_AT]:
    layer.trainable = False

n_unfrozen = sum(1 for l in base_model.layers if l.trainable)
print(f"\nFine Tuning: {n_unfrozen} capas descongeladas")

# Construimos modelo completo para FT
inp_ft = keras.Input(shape=(IM_SIZE, IM_SIZE, 3))
# training=False en base → BatchNorm en modo inferencia
x_ft = base_model(inp_ft, training=False)
x_ft = GlobalAveragePooling2D()(x_ft)
x_ft = Dense(128, activation='relu', name="dense_ft")(x_ft)
x_ft = Dropout(0.5)(x_ft)
out_ft = Dense(1, activation='sigmoid', name="output_ft")(x_ft)
model_ft = keras.Model(inp_ft, out_ft, name="FineTuning_MobileNetV2")

# Transferimos pesos de la cabeza entrenada en TL
model_ft.get_layer("dense_ft").set_weights(head_tl.layers[0].get_weights())
model_ft.get_layer("output_ft").set_weights(head_tl.layers[-1].get_weights())

model_ft.compile(optimizer=keras.optimizers.Adam(learning_rate=1e-5),
                 loss='binary_crossentropy', metrics=['accuracy'])

early_ft = EarlyStopping(monitor='val_loss', patience=3,
                          restore_best_weights=True, verbose=1)

history_ft = model_ft.fit(
    X_train, y_train,
    batch_size=128,
    epochs=8,
    validation_split=0.2,
    callbacks=[early_ft],
    verbose=1
)

# Historial FT
hist_ft = pd.DataFrame(history_ft.history)
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
hist_ft[['loss','val_loss']].plot(ax=axes[0])
axes[0].set_title("Fine Tuning — Loss"); axes[0].set_xlabel("Epoch")
hist_ft[['accuracy','val_accuracy']].plot(ax=axes[1])
axes[1].set_title("Fine Tuning — Accuracy"); axes[1].set_xlabel("Epoch")
plt.tight_layout(); plt.show()

# Evaluación FT
ft_loss, ft_acc = model_ft.evaluate(X_test, y_test, verbose=0)
print(f"\nFine Tuning — Test Loss: {ft_loss:.4f} | Test Accuracy: {ft_acc:.4f}")
probs_ft = model_ft.predict(X_test, verbose=0).flatten()
acc_ft = report_model(probs_ft, y_test, "Fine Tuning (últimas capas MobileNetV2 + LR=1e-5)")

# ============================================================
# 7. COMPARACIÓN DE RESULTADOS
# ============================================================

# CNN ad-hoc práctica anterior: 32×32, 2 bloques Conv-Pool → 67.6%
ACC_CNN_ADHOC = 0.676

comp = pd.DataFrame({
    'Modelo': [
        'CNN ad-hoc (32×32, 2 bloques Conv-Pool)',
        f'Transfer Learning — MobileNetV2 ({IM_SIZE}×{IM_SIZE})',
        f'Fine Tuning — MobileNetV2 ({IM_SIZE}×{IM_SIZE})',
    ],
    'Accuracy (%)': [ACC_CNN_ADHOC * 100, acc_tl * 100, acc_ft * 100]
})

print("\n" + "="*60)
print("  COMPARATIVA DE MODELOS")
print("="*60)
print(comp.to_string(index=False))

fig, ax = plt.subplots(figsize=(9, 4))
bars = ax.bar(comp['Modelo'], comp['Accuracy (%)'],
              color=['#90A4AE', '#2196F3', '#4CAF50'])
ax.set_ylim(50, 100)
ax.set_title("Comparativa — Test Accuracy (%)")
ax.set_ylabel("Accuracy (%)")
for bar in bars:
    ax.text(bar.get_x() + bar.get_width()/2., bar.get_height() + 0.3,
            f"{bar.get_height():.1f}%", ha='center', fontweight='bold')
plt.xticks(rotation=12, ha='right')
plt.tight_layout(); plt.show()

# ============================================================
# EXTRA: TRANSFER LEARNING CON IMAGE AUGMENTATION
# Modelo completo (base congelada) con capas de augmentación.
# ============================================================

base_aug = MobileNetV2(input_shape=(IM_SIZE, IM_SIZE, 3),
                       include_top=False, weights='imagenet')
base_aug.trainable = False

inp_aug  = keras.Input(shape=(IM_SIZE, IM_SIZE, 3))
x_aug    = RandomFlip("horizontal")(inp_aug)
x_aug    = RandomRotation(0.08)(x_aug)
x_aug    = base_aug(x_aug, training=False)
x_aug    = GlobalAveragePooling2D()(x_aug)
x_aug    = Dense(128, activation='relu')(x_aug)
x_aug    = Dropout(0.5)(x_aug)
out_aug  = Dense(1, activation='sigmoid')(x_aug)
model_aug = keras.Model(inp_aug, out_aug, name="TL_MobileNetV2_Augmented")

model_aug.compile(optimizer=keras.optimizers.Adam(1e-3),
                  loss='binary_crossentropy', metrics=['accuracy'])

early_aug = EarlyStopping(monitor='val_loss', patience=3,
                           restore_best_weights=True, verbose=1)

history_aug = model_aug.fit(
    X_train, y_train,
    batch_size=128,
    epochs=8,
    validation_split=0.2,
    callbacks=[early_aug],
    verbose=1
)

hist_aug_df = pd.DataFrame(history_aug.history)
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
hist_aug_df[['loss','val_loss']].plot(ax=axes[0])
axes[0].set_title("[EXTRA] TL + Augmentation — Loss"); axes[0].set_xlabel("Epoch")
hist_aug_df[['accuracy','val_accuracy']].plot(ax=axes[1])
axes[1].set_title("[EXTRA] TL + Augmentation — Accuracy"); axes[1].set_xlabel("Epoch")
plt.tight_layout(); plt.show()

aug_loss, aug_acc = model_aug.evaluate(X_test, y_test, verbose=0)
print(f"\n[EXTRA] TL + Aug — Test Loss: {aug_loss:.4f} | Test Accuracy: {aug_acc:.4f}")
probs_aug = model_aug.predict(X_test, verbose=0).flatten()
acc_aug = report_model(probs_aug, y_test, "[EXTRA] Transfer Learning + Augmentation")

# Tabla comparativa final
comp_final = pd.concat([comp, pd.DataFrame({
    'Modelo': [f'[EXTRA] TL + Aug — MobileNetV2 ({IM_SIZE}×{IM_SIZE})'],
    'Accuracy (%)': [acc_aug * 100]
})], ignore_index=True)
print("\n" + "="*65)
print("  COMPARATIVA FINAL")
print("="*65)
print(comp_final.to_string(index=False))
```
